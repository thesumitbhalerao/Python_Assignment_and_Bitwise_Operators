#### Assignment Operators
##### - Assignment operators are used to assign a value to a variable.
##### - The Basic assignment operator is '='.
##### - They are also used to perform an operation and then assign the result back to the same variable.

##### 1) `=` - Operator is assign a value to a variable
##### Syntax : variable=value

```python
x=10
print("Value of x :",x)
```

##### 2) `+=` - Operator adds a value to the existing variable & assign the result back to the same variable.
##### Syntax : variable += value
##### - It is equivalent to : variable = variable + value

```python
x=10
x+=5
print(x)
```

##### 3) `-=` - Operator subtracts a value to the existing variable & assign the result back to the same variable.
##### Syntax : variable -= value
##### - It is equivalent to : variable = variable - value

```python
x=10
x-=5
print(x)
```

##### 4) `*=` - Operator multiplies the existing value by another value & assign the result back to the same variable.
##### Syntax : variable *= value
##### - It is equivalent to : variable = variable * value

```python
x=10
x*=5
print(x)
```

##### 5) `/=` - Operator divides the existing value by another value & assign the result back to the same variable.
##### Syntax : variable /= value
##### - It is equivalent to : variable = variable / value

```python
x=20
x/=5
print(x)
```

##### 6) `%=` - Operator calculates the remainder & assign the result back to the same variable.
##### Syntax : variable %= value
##### - It is equivalent to : variable = variable % value

```python
x=17
x%=5
print(x)
```

##### 7) `//=` - Operator performs floor division & assign the result back to the same variable.
##### Syntax : variable //= value
##### - It is equivalent to : variable = variable // value

```python
x=17
x//=5
print(x)
```

##### 8) `**=` - Operator raises the variable to the given power & assign the result back to the same variable.
##### Syntax : variable **= value
##### - It is equivalent to : variable = variable ** value

```python
x=2
x**=3
print(x)
```

#### Composite / Compound Operators
##### - Composite opertors are nothing but it is a combination of operators .

```python
x=10
x+=5
print(x)

y=55
y-=5
print(y)

z=60
z*=5
print(z)

a=100
a/=5
print(a)

b=40
b%=3
print(b)

c=35
c//=8
print(c)

d=80
d**=5
print(d)
```
