#### Bitwise Operator : -
##### -Bitwise operators perform operations directly on the individual bits of integer numbers.
##### - A bit can have only two values : 0 -> OFF, 1 -> ON
##### - For example : Decimal 5 -> Binary -> 5=101
##### - Each digit in the binary number represents one bit.

#### Number Systems : -
##### - Computers internally represent data using binary numbers.
##### - Number systems are 1.Decimal, 2.Binary, 3.Octal, 4.Hexadecimal

##### 1 ( AND & )
##### 2 ( OR | )
##### 3 ( XOR ^ )
##### 4 ( NOT ~ )
##### 5 ( LEFT SHIFT << )
##### 6 ( RIGHT SHIFT >> )

#### 1) and & : 
#### 5 = 0101
#### 3 = 0011
#### r = 0001 = 1

```python
a=5
b=3
result=a&b
print(result)
```

#### 2) or | : 
#### 5 = 0101
#### 3 = 0011
#### r = 0111 = 7

```python
a=5
b=3
result=a|b
print(result)
```

#### 3) XOR ^ : 
#### 5 = 0101
#### 3 = 0011
#### r = 0110 = 6

```python
a=5
b=3
result= a^b
print(result)
```

#### 4) Not ~ : 
#### 5 = 0101 
#### 3 = 0011 
#### formula ~n = -(n+1)
#### ~5 = -(5+1) -> -6
#### ~3 = -(3+1) -> -4

```python
a=5
b=3
print(~a)
print(~b)
```

#### 5) left shift << : 

```python
print(5<<1)
print(7<<3)  ### this is the left shifting
```

#### 6) right shift >> : 

```python
print(5>>1) ### this is the right shifting
print(7>>3)
```
